```python
# kasir_faktur.py
# Program Generator Struk Transaksi & Kasir Otomatis

# ==========================================
# 1. INPUT DATA DARI PENGGUNA
# ==========================================
print("=== INPUT TRANSAKSI ===")
nama_pelanggan = input("Masukkan Nama Pelanggan             : ")
total_awal = float(input("Total Belanja Sebelum Diskon (Rp)   : "))
persentase_diskon = float(input("Persentase Diskon (%)    : "))

# ==========================================
# 2. PERHITUNGAN
# ==========================================
nominal_diskon = total_awal * (persentase_diskon / 100)
total_bayar = total_awal - nominal_diskon

# Format nominal ke mata uang: ribuan koma (,) dan 2 angka desimal (.2f)
str_total_awal = f"Rp{total_awal:,.2f}"
str_diskon = f"Rp{nominal_diskon:,.2f}"
str_total_bayar = f"Rp{total_bayar:,.2f}"

# Label diskon dinamis (menampilkan angka persentase)
label_diskon = f"Diskon ({persentase_diskon}%)"

# ==========================================
# 3. CETAK STRUK DENGAN F-STRING ALIGNMENT
# ==========================================
LEBAR = 36  # Lebar total struk sesuai spesifikasi

print("\n" + "=" * LEBAR)
# Header dan Nama Pelanggan rata tengah ( ^ ) dengan lebar 36
print(f"{'STRUK PEMBAYARAN':^{LEBAR}}")
print(f"{nama_pelanggan:^{LEBAR}}")
print("=" * LEBAR)

# Label rata kiri (< 18) dan Nominal rata kanan (> 18) -> Total = 36
print(f"{'Total Awal':<18}{str_total_awal:>18}")
print(f"{label_diskon:<18}{str_diskon:>18}")
print("-" * LEBAR)
print(f"{'Total Bayar':<18}{str_total_bayar:>18}")
print("=" * LEBAR)
print(f"{'TERIMA KASIH':^{LEBAR}}")
print("=" * LEBAR)
```
