```python
# Program: konversi_ipk.py
# Deskripsi: Memproses transkrip nilai akademik dari input string satu baris

# 1. Menerima input satu baris
data_input = input("Masukkan data (Nama, NIM, Nilai Tugas, Nilai UTS, Nilai UAS): ")

# Memisahkan string berdasarkan tanda koma ','
bagian = data_input.split(",")

# Menyimpan ke dalam variabel dan membersihkan spasi berlebih dengan .strip()
nama = bagian[0].strip()
NIM = bagian[1].strip()

# 2. Konversi ketiga nilai ke bentuk float
nilai_tugas = float(bagian[2].strip())
nilai_uts = float(bagian[3].strip())
nilai_uas = float(bagian[4].strip())

# Bobot penilaian
bobot_tugas = 0.20
bobot_uts = 0.35
bobot_uas = 0.45

# 3. Menghitung Nilai Akhir
nilai_akhir = (
    (nilai_tugas * bobot_tugas)
    + (nilai_uts * bobot_uts)
    + (nilai_uas * bobot_uas)
)

# 4. Menampilkan tabel ringkasan dalam 1 frame kotak
# Menggunakan format :.0% untuk bobot dan :.1f untuk angka 1 desimal
garis_pemisah = "+" + "-" * 20 + "+" + "-" * 10 + "+" + "-" * 12 + "+"

print("\n" + garis_pemisah)
print(f"| {'TRANSKRIP NILAI MAHASISWA':^43}|")
print(garis_pemisah)
print(f"| Nama : {nama:<35} |")
print(f"| NIM  : {NIM:<35} |")
print(garis_pemisah)
print(f"| {'Komponen':<18} | {'Bobot':<8} | {'Nilai':<10} |")
print(garis_pemisah)
print(f"| {'Tugas':<18} | {bobot_tugas:<8.0%} | {nilai_tugas:<10.1f} |")
print(f"| {'UTS':<18} | {bobot_uts:<8.0%} | {nilai_uts:<10.1f} |")
print(f"| {'UAS':<18} | {bobot_uas:<8.0%} | {nilai_uas:<10.1f} |")
print(garis_pemisah)
print(f"| {'Nilai Akhir':<18} | {1.0:<8.0%} | {nilai_akhir:<10.1f} |")
print(garis_pemisah)
```
